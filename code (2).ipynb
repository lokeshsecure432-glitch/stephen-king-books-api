#  Stephen King Books Data Extraction & Processing

## 1. Import Required Libraries

We import the required Python libraries for API requests, JSON processing,
data manipulation, and program control.

```python
import json 
import pandas as pd 
import sys
import requests
```

## 2. Fetch Data from API and Save as JSON

In this section, we fetch book data from the **Stephen King Books REST API** using the `requests` library.

###  *Steps Performed*

1. **Send an HTTP GET request** to the API using `requests.get()`.
2. **Check the response status code** to verify whether the data was fetched successfully.
3. **Convert the API response into Python data** using `res.json()`.
4. **Save the API response** into a `book_data.json` file using `json.dump()`.
5. Use `f.seek(0)` to **move the file pointer back to the beginning** of the file.
6. Use `json.load()` to **read the JSON data from the file** and store it in `json_data`.
7. Finally, display the **top-level keys** of the JSON data using `json_data.keys()`.

###  *Why This Is Done*

This demonstrates the complete flow of:

**REST API → HTTP Response → Python Data → JSON File → Python Data**

The JSON file allows us to **store the raw API data locally** so that it can be used later for data processing and analysis.


```python
url = 'https://stephen-king-api.onrender.com/api/books'

res = requests.get(url)  

if res.status_code != 200 :
    sys.exit('data is not fetched successfully')   

#---------------direct only store in variable -------------------
json_data = res.json()   

#----------save and used save data into file ------------

with open('book_data.json','w+',encoding ='utf-8') as f:
    json.dump(res.json(),f,indent = 4) 

    f.seek(0)   

    json_data = json.load(f)   

print(f'the key of this json_data : {list(json_data.keys())}')

```

## 3. Convert JSON Data into Pandas DataFrame

Convert the JSON book data into a **Pandas DataFrame** using `pd.json_normalize()`.

```python
df  = pd.json_normalize(json_data['data'])  # used to convert nested data into tabular form like dataframe 
df.columns
```

## 4. Select Required Columns

Select the required columns and display the first 5 rows.

```python
selected_df = df[['id', 'Year', 'Title', 'Publisher']]
selected_df.head()
```

## 5. Export Data to CSV

Save the selected DataFrame as a **CSV file** without including the index.

```python
selected_df.to_csv('book_data.csv',index = False)
```
